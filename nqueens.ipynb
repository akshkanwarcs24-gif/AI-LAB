```python
import random

def calculate_heuristic(state):
    """Calculates the number of attacking pairs of queens."""
    n = len(state)
    attacking_pairs = 0
    for i in range(n):
        for j in range(i + 1, n):
            if state[i] == state[j]:
                attacking_pairs += 1
            elif abs(state[i] - state[j]) == abs(i - j):
                attacking_pairs += 1
    return attacking_pairs

def get_neighbors(state):
    """Generates all neighboring states by moving one queen at a time within its column."""
    neighbors = []
    n = len(state)
    for i in range(n):
        for val in range(n):
            if val != state[i]:
                neighbor = list(state)
                neighbor[i] = val
                neighbors.append(neighbor)
    return neighbors

def print_board(state, state_number):
    """Prints the board state as an ASCII matrix with dashes for blank squares and the exact state number processed."""
    print(f"--- State {state_number} ---")
    n = len(state)
    for r in range(n):
        row_str = ""
        for c in range(n):
            if state[c] == r:
                row_str += " Q "
            else:
                row_str += " - "
        print(row_str)
    print()

def hill_climbing_n_queens(n, initial_state=None, max_sideways_moves=100):
    """Solves N-Queens using hill climbing, allowing sideways moves to avoid local plateaus."""
    if initial_state is not None:
        current_state = list(initial_state)
    else:
        current_state = [random.randint(0, n - 1) for _ in range(n)]

    current_h = calculate_heuristic(current_state)
    states_processed = 1
    sideways_count = 0

    print(f"Initial State: {current_state} (h = {current_h})")
    print_board(current_state, states_processed)

    while True:
        neighbors = get_neighbors(current_state)
        best_neighbors = []
        best_neighbor_h = current_h


        for neighbor in neighbors:
            states_processed += 1
            h = calculate_heuristic(neighbor)
            if h < best_neighbor_h:
                best_neighbors = [neighbor]
                best_neighbor_h = h
            elif h == best_neighbor_h:
                best_neighbors.append(neighbor)


        if best_neighbor_h > current_h:
            break

        if best_neighbor_h == current_h:
            if sideways_count >= max_sideways_moves or not best_neighbors:
                break
            sideways_count += 1
            current_state = random.choice(best_neighbors)
            print(f"Sideways Move {sideways_count} (h remains {current_h}): {current_state}")
        else:

            sideways_count = 0
            current_state = random.choice(best_neighbors)
            current_h = best_neighbor_h
            print(f"Moved to State (h decreases to {current_h}): {current_state}")
            print_board(current_state, states_processed)

        if current_h == 0:
            break

    if current_h == 0:
        print(f"\nSuccess! Global minimum reached.")
    else:
        print(f"\nStuck at Local Optimum (Failure to find global solution).")

    return current_state, current_h, states_processed

n = 6
initial_state = [0, 1, 2, 3, 4, 5]

final_state, final_h, total_states = hill_climbing_n_queens(n, initial_state=initial_state)

print(f"Final State: {final_state}")
print(f"Final Heuristic (Attacking Pairs): {final_h}")
print(f"Total States Processed: {total_states}")
```
