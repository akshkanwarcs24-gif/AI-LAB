```python
import heapq
import copy
class PuzzleNode:
    def __init__(self, state, parent=None, action=None, g=0, h=0):
        self.state = state
        self.parent = parent
        self.action = action
        self.g = g
        self.h = h
        self.f = g + h
    def __lt__(self, other):
        return self.f < other.f
    def __eq__(self, other):
        return self.f == other.f
def get_blank_position(state):
    for r in range(3):
        for c in range(3):
            if state[r][c] == 0:
                return r, c
def get_neighbors(node, goal_state):
    neighbors = []
    r, c = get_blank_position(node.state)
    moves = {
        'UP': (r - 1, c),
        'DOWN': (r + 1, c),
        'LEFT': (r, c - 1),
        'RIGHT': (r, c + 1)
    }
    for action, (nr, nc) in moves.items():
        if 0 <= nr < 3 and 0 <= nc < 3:
            next_state = [list(row) for row in node.state]
            next_state[r][c], next_state[nr][nc] = next_state[nr][nc], next_state[r][c]
            next_state_tuple = tuple(tuple(row) for row in next_state)
            h = manhattan_distance_heuristic(next_state_tuple, goal_state)
            neighbors.append(PuzzleNode(next_state_tuple, node, action, node.g + 1, h))
    return neighbors

def manhattan_distance_heuristic(state, goal_state):
    goal_positions = {}
    for r in range(3):
        for c in range(3):
            goal_positions[goal_state[r][c]] = (r, c)
    distance = 0
    for r in range(3):
        for c in range(3):
            val = state[r][c]
            if val != 0:
                gr, gc = goal_positions[val]
                distance += abs(r - gr) + abs(c - gc)
    return distance
def reconstruct_path(node):
    path = []
    curr = node
    while curr:
        path.append((curr.action, curr.state))
        curr = curr.parent
    return path[::-1]
def solve_a_star(start_state, goal_state):
    start_tuple = tuple(tuple(row) for row in start_state)
    goal_tuple = tuple(tuple(row) for row in goal_state)
    start_node = PuzzleNode(
        state=start_tuple,
        g=0,
        h=manhattan_distance_heuristic(start_tuple, goal_tuple)
    )
    open_set = []
    heapq.heappush(open_set, start_node)
    closed_set = set()
    nodes_expanded = 0
    while open_set:
        current_node = heapq.heappop(open_set)
        nodes_expanded += 1
        if current_node.state == goal_tuple:
            return reconstruct_path(current_node), nodes_expanded
        closed_set.add(current_node.state)
        for neighbor in get_neighbors(current_node, goal_tuple):
            if neighbor.state in closed_set:
                continue
            heapq.heappush(open_set, neighbor)
    return None, nodes_expanded
def print_state(state):
    for row in state:
        print(" ".join(str(val) if val != 0 else "_" for val in row))
    print()
initial_state = [
    [1, 2, 3],
    [5, 0, 6],
    [4, 7, 8]
]
goal_state = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 0]
]
print("Initial State:")
print_state(initial_state)
print("Goal State:")
print_state(goal_state)
path, nodes_expanded = solve_a_star(initial_state, goal_state)
if path:
    print(f"Solvable! Steps taken: {len(path) - 1}")
    print(f"Nodes expanded: {nodes_expanded}\n")
    goal_tuple = tuple(tuple(row) for row in goal_state)
    for step, (action, state) in enumerate(path):
        g = step
        h = manhattan_distance_heuristic(state, goal_tuple)
        f = g + h
        if action:
            print(f"Step {step}: Move blank {action} | g(n)={g}, h(n)={h}, f(n)={f}")
        else:
            print(f"Start State | g(n)={g}, h(n)={h}, f(n)={f}")
        print_state(state)
else:
    print("No solution found.")
```
