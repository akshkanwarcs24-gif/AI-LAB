```python
class Symbol:
    def __init__(self, name):
        self.name = name
    def evaluate(self, model):
        return model.get(self.name, False)
    def __repr__(self):
        return self.name

class Not:
    def __init__(self, op):
        self.op = op
    def evaluate(self, model):
        return not self.op.evaluate(model)
    def __repr__(self):
        return f"~({self.op})"

class And:
    def __init__(self, left, right):
        self.left = left
        self.right = right
    def evaluate(self, model):
        return self.left.evaluate(model) and self.right.evaluate(model)
    def __repr__(self):
        return f"({self.left} & {self.right})"

class Or:
    def __init__(self, left, right):
        self.left = left
        self.right = right
    def evaluate(self, model):
        return self.left.evaluate(model) or self.right.evaluate(model)
    def __repr__(self):
        return f"({self.left} | {self.right})"

class Implies:
    def __init__(self, left, right):
        self.left = left
        self.right = right
    def evaluate(self, model):
        return (not self.left.evaluate(model)) or self.right.evaluate(model)
    def __repr__(self):
        return f"({self.left} => {self.right})"


def get_symbols(expression):
    if isinstance(expression, Symbol):
        return {expression.name}
    elif isinstance(expression, Not):
        return get_symbols(expression.op)
    elif isinstance(expression, (And, Or, Implies)):
        return get_symbols(expression.left).union(get_symbols(expression.right))
    return set()

def evaluate_kb(kb, model):
    for clause in kb:
        if not clause.evaluate(model):
            return False
    return True

def check_all(kb, query, symbols, model):
    if not symbols:
        if not evaluate_kb(kb, model):
            return True
        else:
            return query.evaluate(model)
    else:
        symbol = symbols[0]
        remaining = symbols[1:]
        return (check_all(kb, query, remaining, {**model, symbol: True}) and
                check_all(kb, query, remaining, {**model, symbol: False}))

def tt_entails(kb, query):
    symbols = set()
    for clause in kb:
        symbols.update(get_symbols(clause))
    symbols.update(get_symbols(query))
    return check_all(kb, query, list(symbols), {})

# Helper to dynamically construct expressions
def parse_expression(expr_str, symbols_dict):
    expr_str = expr_str.strip()
    # Simple parser mapping standard syntax strings to class structures using eval
    # We define local environment variables for evaluation
    safe_env = {
        'Not': Not,
        'And': And,
        'Or': Or,
        'Implies': Implies
    }
    for sym_name, sym_obj in symbols_dict.items():
        safe_env[sym_name] = sym_obj
    try:
        return eval(expr_str, {"__builtins__": None}, safe_env)
    except Exception as e:
        raise ValueError(f"Could not parse expression '{expr_str}'. Error: {e}")

# Dynamic Input from User
print("=== Truth Table Entailment Setup ===")
print("First, define the symbols you want to use (comma-separated, e.g., A, B, C):")
symbols_input = input("Symbols: ")
symbols_list = [s.strip() for s in symbols_input.split(",") if s.strip()]
symbols_dict = {name: Symbol(name) for name in symbols_list}

print("\nNow, write expressions using Python classes: SymbolName, Not(x), And(x, y), Or(x, y), Implies(x, y)")
print("Example KB input: Implies(A, B), A")
kb_input = input("Enter Knowledge Base clauses (comma-separated): ")
try:
    # Split by comma outside parentheses
    import re
    kb_clauses_str = re.split(r',\s*(?![^()]*\))', kb_input)
    kb = [parse_expression(clause, symbols_dict) for clause in kb_clauses_str if clause.strip()]

    query_input = input("Enter Query expression: ")
    query = parse_expression(query_input, symbols_dict)

    print(f"\nKB: {kb}")
    print(f"Query: {query}")

    result = tt_entails(kb, query)
    print(f"Does KB entail the query '{query}'? {result}")
except Exception as e:
    print(f"Error processing inputs: {e}")
```
