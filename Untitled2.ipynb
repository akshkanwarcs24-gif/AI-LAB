```python
import itertools

GOAL_STATE = (1, 2, 3, 4, 5, 6, 7, 8, 0)

def get_neighbors(state):
    """Generates all valid next states by moving the blank space (0)."""
    neighbors = []
    zero_idx = state.index(0)
    row, col = divmod(zero_idx, 3)

    moves = []
    if row > 0: moves.append(-3) # Up
    if row < 2: moves.append(3)  # Down
    if col > 0: moves.append(-1) # Left
    if col < 2: moves.append(1)  # Right

    for move in moves:
        new_idx = zero_idx + move
        new_state = list(state)
        new_state[zero_idx], new_state[new_idx] = new_state[new_idx], new_state[zero_idx]
        neighbors.append(tuple(new_state))

    return neighbors

def depth_limited_dfs(state, goal, depth, path):
    """Helper DFS that stops expanding if the current depth limit is reached."""
    if state == goal:
        return path
    if depth <= 0:
        return None

    for neighbor in get_neighbors(state):
        if neighbor not in path: # Avoid cyclic paths in the current branch
            result = depth_limited_dfs(neighbor, goal, depth - 1, path + (neighbor,))
            if result is not None:
                return result
    return None

def iddfs_solver(start_state):
    """Core IDDFS function incrementing depth thresholds dynamically."""
    if start_state == GOAL_STATE:
        return [start_state]

    for depth in itertools.count(1):
        path = depth_limited_dfs(start_state, GOAL_STATE, depth, (start_state,))
        if path is not None:
            return path

def print_board(state):
    """Formats a flat 9-element tuple into a 3x3 grid layout."""
    for i in range(0, 9, 3):
        print(f"{state[i]} {state[i+1]} {state[i+2]}")
    print("-" * 5)

if __name__ == "__main__":
    initial_state = (1, 2, 3, 4, 0, 6, 7, 5, 8)

    print("Initial State:")
    print_board(initial_state)

    print("Searching for optimal solution via IDDFS...")
    solution_path = iddfs_solver(initial_state)

    if solution_path:
        print(f"\nSolution found in {len(solution_path) - 1} moves!\n")
        for step, state in enumerate(solution_path):
            print(f"Step {step}:")
            print_board(state)
    else:
        print("No solution could be found.")

```
